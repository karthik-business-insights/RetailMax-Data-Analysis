data cleaning


```python
import numpy as np
import pandas as pd
df=pd.read_excel("RetailMax e-commerce.xlsx")
pd.set_option('display.max_columns',None)
pd.set_option('display.width',None)
display(df.head(10))
df.info()
df.isnull().sum()
print("duplicated:" ,df.duplicated().sum())
df["ProductName"].unique()
df["OrderDate"] = pd.to_datetime(df["OrderDate"])
df["ShipDate"] = pd.to_datetime(df["ShipDate"])
df["ShipDate"].info()

df.duplicated().sum()
df.columns = df.columns.str.strip()

text_columns = df.select_dtypes(include="object").columns
df[text_columns] = df[text_columns].apply(lambda x: x.str.strip())
 # checking invalid numbers
(df["Quantity"] <= 0).sum()
print((df["ShipDate"] <= df["OrderDate"]).sum())
expected_sales = df["Quantity"] * df["UnitPrice"] * (1 - df["Discount"] / 100)

mismatch = df[expected_sales.round(2) != df["Sales"].round(2)]

mismatch[["Quantity", "UnitPrice", "Discount", "Sales"]].head()
len(mismatch)
mismatch.index
expected_sales = (
    df["Quantity"] * df["UnitPrice"] * (1 - df["Discount"] / 100)
).round(2)

mask = expected_sales != df["Sales"].round(2)

df.loc[mask, "Sales"] = expected_sales[mask]
expected_sales = (
    df["Quantity"] * df["UnitPrice"] * (1 - df["Discount"] / 100)
).round(2)

(expected_sales == df["Sales"].round(2)).all()
(df["Profit"] < 0).sum()
expected_profit = df["Sales"] - (df["CostPrice"] * df["Quantity"])

(expected_profit.round(2) == df["Profit"].round(2)).all()
expected_profit = df["Sales"] - (df["CostPrice"] * df["Quantity"])

mismatch = df[expected_profit.round(2) != df["Profit"].round(2)]

mismatch[[
    "OrderID",
    "Quantity",
    "Sales",
    "CostPrice",
    "Profit"
]]
expected_profit = df["Sales"] - (df["CostPrice"] * df["Quantity"])

df["ExpectedProfit"] = expected_profit.round(2)
df["Difference"] = (df["Profit"] - df["ExpectedProfit"]).round(2)

mismatch = df[df["Difference"] != 0]

mismatch[[
    "OrderID",
    "Quantity",
    "Sales",
    "CostPrice",
    "Profit",
    "ExpectedProfit",
    "Difference"
]]
len(mismatch)
mismatch.shape[0]
expected_profit = df["Sales"] - (df["CostPrice"] * df["Quantity"])

mask = expected_profit.round(2) != df["Profit"].round(2)

df.loc[mask, "Profit"] = expected_profit[mask].round(2)
expected_profit = df["Sales"] - (df["CostPrice"] * df["Quantity"])

(expected_profit.round(2) == df["Profit"].round(2)).all()
df["Year"] = df["OrderDate"].dt.year
df["Month"] = df["OrderDate"].dt.month_name()
df["Quarter"] = df["OrderDate"].dt.quarter
df["ShippingDays"] = (df["ShipDate"] - df["OrderDate"]).dt.days
df["ProfitMargin"] = (df["Profit"] / df["Sales"]) * 100
df["Difference"] = df["ExpectedProfit"] - df["Profit"]
df["Sales"] = df["Sales"].round(2)
df["Profit"] = df["Profit"].round(2)
df["ProfitMargin"] = df["ProfitMargin"].round(2)
# Day of the week
df["DayOfWeek"] = df["OrderDate"].dt.day_name()

# Weekend orders
df["IsWeekend"] = df["DayOfWeek"].isin(["Saturday", "Sunday"])
display(df.head(1))
df.to_excel("RetailMax_Cleaned.xlsx", index=False)
months_to_remove = [
    "August",
    "September",
    "October",
    "November",
    "December"
]

df = df[
    ~(
        (df["Year"] == 2026) &
        (df["Month"].isin(months_to_remove))
    )
]
display(df.head(10))
df[df["Year"]==2026]["Month"].unique()
```
